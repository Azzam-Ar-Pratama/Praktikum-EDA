```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Tugas Analisis 2:
Kolom yang memiliki data kosong (None) adalah:
1. Kolom menu: baris ke-5 kosong.
2. Kolom terjual: baris ke-3 kosong.

Jika data langsung dianalisis tanpa dibersihkan, hasil analisis bisa tidak akurat.

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Tugas Analisis 3:

Dari hasil df.info(), kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total (110) adalah:

1. menu: 107 non-null, berarti ada 3 data kosong.
2. terjual: 104 non-null, berarti ada 6 data kosong.

Artinya, pada kolom tersebut terdapat missing value (data yang kosong) sehingga perlu ditangani terlebih dahulu sebelum dilakukan analisis agar hasilnya akurat.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Tugas Analisis 4:

Kolom terjual diisi dengan 0 karena jika jumlah penjualan kosong, kita dapat menganggap tidak ada data penjualan atau belum ada yang terjual, sehingga masih bisa digunakan dalam analisis.

Sedangkan baris dengan menu kosong dihapus karena nama menu merupakan informasi penting untuk mengetahui barang yang dijual. Jika nama menunya tidak diketahui, data tersebut sulit digunakan untuk menganalisis penjualan berdasarkan menu.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Tugas Analisis 5:

Sebelum drop_duplicates(), jumlah baris dataset adalah 110 baris.

Setelah menjalankan:

df = df.drop_duplicates()

jumlah baris menjadi 110 dikurangi jumlah baris yang duplikat.
Jadi, jika hasil:

print(df.duplicated().sum())

misalnya 6, maka:

110 − 6 = 104 baris.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu yang menghasilkan total_pendapatan tertinggi dapat diketahui dari nilai terbesar pada hasil groupby().
